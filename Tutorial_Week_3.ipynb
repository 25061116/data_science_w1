```python
import numpy as np
#1
array_2x2 = np.array([[1,2], [3,4]])
print(array_2x2)
print(f"shape: {array_2x2.shape}\n")
```

```python
#2
array_3x4 = np.random.rand(3, 4)
print(array_3x4)
print(f"shape: {array_3x4.shape}\n")
```

```python
#3
ones_array = np.ones((3,3))
zeros_array = np.zeros((2,4))


print("zeroes array (2x4):")
print(zeros_array)
print("ones array (2x4):")
print(ones_array)
print(f"ones shape: {ones_array.shape}, zeroes shape: {zeros_array.shape}\n")

```

```python
#4
random_3x6 = np.random.rand(3, 6)
transposed = random_3x6.T
print("original (3x6)")
print(random_3x6)
print("transported (6x3)")
print(transposed)
print(f"shape: {random_3x6.shape}, {transposed.shape}\n")

```

```python
#5
n, p, m, = 4, 3, 5,
array1 = np.random.rand(n , p)
array2 =  np.random.rand(p , m)
dot_product = np.dot(array1,array2)

print(f"Array 1 ({n}x{p}):")
print(array1)
print(f"array 2 ({p}x{m})")
print(array2)
print(f"Dot product result ({n}x{m})")
print(dot_product)
print(f"Dot product shape: {dot_product.shape}")
```
